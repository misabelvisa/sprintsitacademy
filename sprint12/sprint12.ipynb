**Nivell 1**

Des d’un Jupyter Notebook faràs els següents exercicis utilitzant la llibreria requests de Python.

Exploració bàsica amb una API de laboratori.
*1. Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir:*

Llista de publicacions (/posts)
Llista d’usuaris (/users)
Llista de tasques (/todos)

```python
import requests

url = "https://jsonplaceholder.typicode.com/"

#Llista de publicacions (/posts)
posts = requests.get(f"{url}/posts")
print(posts.json())

#Llista d’usuaris (/users)
users = requests.get(f"{url}/users")
print(users.json())

#Llista de tasques (/todos)
todos = requests.get(f"{url}/todos")
print(todos.json())
```

*2. Mostra per pantalla:*

La quantitat total de cada recurs.
El codi d'estat de cada petició.

```python
#La quantitat total de cada recurs.
print("Total posts:", len(posts.json()))
print("Total users:", len(users.json()))
print("Total todos:", len(todos.json()))

#El codi d'estat de cada petició.
print("Status_code posts:", posts.status_code)
print("Status_code users:", users.status_code)
print("Status_code todos:", todos.status_code)
```

*3. Fes una petició a una publicació inexistent per obtenir un error 404 i mostra el codi d'estat rebut.*

```python
error_404 = requests.get(f"{url}/posts/119")
print("Status_code error:", error_404.status_code)
```

*4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.*

Mostra la resposta JSON.
Mostra el codi d'estat.

```python
publicacion_ficticia = {"title": "Título fictício",
                        "body": "Este es un texto fictício",
                        "userId": 1}

nuevo_post = requests.post(f"{url}/posts", json=publicacion_ficticia)

#Mostra la resposta JSON.
print("Respuesta JSON:", nuevo_post.json())

#Mostra el codi d'estat.
print("Status_code post:", nuevo_post.status_code)
```

*5. Fes una petició PATCH per modificar parcialment una publicació existent.*

Mostra la resposta JSON.
Mostra el codi d'estat.

```python
modificacion = {"title": "Título modificado parcialmente"}

response_patch = requests.patch(f"{url}/posts/1", json=modificacion)

#Mostra la resposta JSON.
print("Respuesta JSON:", response_patch.json())

#Mostra el codi d'estat.
print("Status_code modificación:", response_patch.status_code)
```

*6. Fes una petició DELETE sobre una publicació.*

Mostra la resposta JSON.
Mostra el codi d'estat.

```python
response_delete = requests.delete(f"{url}/posts/1")

#Mostra la resposta JSON.
print("Respuesta JSON:", response_delete.json())

#Mostra el codi d'estat.
print("Status_code modificación:", response_delete.status_code)
```

*3. Fes una petició GET senzilla:*

Mostra el codi d'estat de la resposta.
Imprimeix de forma clara alguns dels camps de la resposta JSON.

```python
url = "https://open.er-api.com/v6/latest/EUR"

response = requests.get(url)

# 1. Mostrar el status_code
print("Status_code:", response.status_code)

# 2. Convertir a JSON
data = response.json()

# 3. Mostrar algunos campos
print("Resultado:", data["result"])
print("Base:", data["base_code"])
print("Data actualitzación:", data["time_last_update_utc"])
print("Tipo de cambio EUR a USD:", data["rates"]["USD"])
```

*4. Converteix les dades a un DataFrame de pandas:*

Mostra les primeres files del DataFrame.

```python
import pandas as pd

rates = data["rates"]

df = pd.DataFrame.from_dict(rates, orient="index", columns=["rate_vs_EUR"])
#crear toda la matriz de tipos de cambio
currencies = df.index
tipos_cambio = pd.DataFrame(index=currencies, columns=currencies)

for c1 in currencies:
    for c2 in currencies:
        tipos_cambio.loc[c1, c2] = rates[c2] / rates[c1]

#convertir a float
tipos_cambio = tipos_cambio.astype(float)

tipos_cambio.head()
```

API d'Open Data Barcelona

*1. Utilitza l’API d’Open Data BCN per:*

Cercar un dataset del teu interès mitjançant package_search i package_show.

```python
import requests

url_search = "https://opendata-ajuntament.barcelona.cat/data/api/action/package_search"

response = requests.get(url_search)
data = response.json()

print("Total encontrados:", data["result"]["count"])

for ds in data["result"]["results"]:
    print("-", ds["title"], "| ID:", ds["id"])
```

```python
url_show = "https://opendata-ajuntament.barcelona.cat/data/api/action/package_show"
params = {"id": "28c20408-5bce-41db-9b83-1b85ac9b2548"}

response = requests.get(url_show, params=params)
population = response.json()

#mostrar título y descripción
print("Título:", population["result"]["title"])
print("Descripción:", population["result"]["notes"])
population
```

*2. Dels resultats obtinguts, selecciona’n un que tingui recursos disponibles en CSV o JSON.*

```python
resources = population["result"]["resources"]

datos_csv = [recurso for recurso in resources if recurso["format"].lower() in ["csv"]]

print("Recursos CSV encontrados:")
for recurso in datos_csv:
    print("-", recurso["name"], "| formato:", recurso["format"], "| URL:", recurso["url"])
```

```python
#url datos a utilizar:
url_recurso = datos_csv[0]["url"]
```

*3. Amb el resource_id del recurs seleccionat, realitza una consulta mitjançant datastore_search per:*

Recuperar almenys 100 registres.

```python
recurso_id = datos_csv[0]["id"]
print(recurso_id)
```

```python
url_datastore = "https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search"

params = {"resource_id": recurso_id,
          "limit": 100}

response = requests.get(url_datastore, params=params)
datos_100 = response.json()

print("Registros recuperados:", len(datos_100["result"]["records"]))
```

*4. Converteix els resultats a un DataFrame.*

```python
import pandas as pd

datos_nacionalidad_100 = pd.DataFrame(datos_100["result"]["records"])
datos_nacionalidad_100.head()
```

*5. Desa el DataFrame en un fitxer .csv.*

```python
datos_nacionalidad_100.to_csv("datos_nacionalidad_100.csv", index=False, encoding='utf-8')
print("csv guardado con éxito.")
```
